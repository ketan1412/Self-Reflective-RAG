```python
# from langchain.vectorstores import FAISS
from langchain_community.vectorstores import FAISS
# from langchain.embeddings import CohereEmbeddings
from langchain_cohere import CohereEmbeddings
# from langchain.schema import Document
from langchain_core.documents import Document
import faiss, pickle

Query=""" RAG - Retrieval systems fetch relevant content that is used by LLMs to produce sensible output rather than generating a content of its own(hallucination).
Without retrieval systems, models will become less up-to-date.
The models need to access up-to-date data, that is possible using retrieval systems.
Retrieval system example:
Facebook AI Similarity Search-n developed by Meta.

RAG - Retriever + Generator
Retriever gets latest upto date information and sends it to LLMs for generating the final output.

Customize - Models parameters are updated according to the domain specific data.
This concept is similar to transfer learning in neural networks.
Weights of the only last few layers of the neural network are updated not all the layers to enter organization data.
How RAG Works:
1. Input Query: The user provides a query.
2. Retrieval: The retriever fetches relevant documents or passages from a knowledge source. Document loader send documents for embedding
3. Augmentation: The retrieved context is passed to the generator, often concatenated with the original query.
4. Generation: The generator produces a final response based on both the query and the retrieved context. (LLM)
"""
```

```python
#Converting text into Documents format
doc = Document(page_content=Query)

# Embedding model
embedding_model = CohereEmbeddings(model="embed-english-v3.0", user_agent="langchain", client="cohere", cohere_api_key="API_KEY")

# Converting Documents into embeddings using cohere embeddings and storing embeddings in the vectore data base
vector_store = FAISS.from_documents(embedding=embedding_model, documents=[doc])

# saving the embeddings on a local drive for re-use
vector_store.save_local("content/sample_data")

# Loading the embeddings
vector_store_loaded = vector_store.load_local("content/sample_data", embeddings=embedding_model, allow_dangerous_deserialization=True)
```
