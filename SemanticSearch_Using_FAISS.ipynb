```python
import faiss, cohere, numpy as np
```

```python
text = [
    "RAG combines retrieval with generation for better answers.",

    "Qdrant is an open-source vector database with a free cloud tier.",

    "FAISS is a local vector store useful for prototyping RAG pipelines.",

    '''ChromaDB is a lightweight, open-source vector database designed for AI applications. It's used to store, manage, and search for vector embeddings, which are numerical
    representations of unstructured data like text or images. This allows developers to perform fast similarity searches and build applications like chatbots, semantic
    search engines, and recommendation systems.''',

    "LangChain integrates Qdrant for RAG pipelines.",

    "Pinecone and Weaviate are other vector DBs.",

    "Vector databases store embeddings for search.",

    "Linear Regression is used to find relationship between the target variable -dependent variable and the undependent variable. It looks for straight line relationship",

    " Embeddings are indexed, stored, retrived from a vector database",

    '''A vector database is a type of database that stores data as high-dimensional vectors, which are numerical representations of things like text or images. It's built
    for fast and efficient similarity searches, allowing it to quickly find data points that are mathematically "similar" to a query. This makes it a crucial tool for AI
    applications like semantic search and recommendation systems.''',

    '''A relational database is a type of database that stores and organizes data in a structured way, using tables to represent data and its relationships.
    Think of it like a collection of spreadsheets where each spreadsheet is a table.
    Tables: Data is organized into tables, also known as "relations." Each table has a specific purpose, for example, a "Customers" table or an "Orders" table.
    Rows and Columns: Each table consists of rows (records or "tuples") and columns (attributes). A row represents a single entry (e.g., one customer),
     while a column holds a specific piece of information about that entry (e.g., the customer's name).
    Relationships: The "relational" part comes from the ability to link data between different tables.
    This is done using keys, which create a logical connection that makes it easy to combine data from multiple tables.''',

    '''A vector database is a specialized database designed to efficiatly store and manage high-dimensional embeddings.
    These embeddings are dense numerical representations, meaning they contain very few zero values. Their purpose is to capture the contextual meaning of data, especially text.'''
  ]
```

```python
co = cohere.Client("API_KEY")

emb = co.embed(texts=text, model="embed-v4.0").embeddings

# FAISS expects raw float32 arrays

# FAISS internally uses C++ arrays (float*) for computations
# Numpy arrays with dtype=float32 are the closest representation in python

# if you pass a list or a tensor, FAISS can not process it.

embeddings = np.array(emb).astype('float32')

shape = embeddings.shape[1]
shape

index = faiss.IndexFlatL2(embeddings.shape[1])
index.add(embeddings)

query = ['What is vector database ?']

emb_query = co.embed(texts=query, model="embed-v4.0").embeddings

embeddings_query = np.array(emb_query).astype('float32')

distance, indices = index.search(embeddings_query, k=2)

# to make a string bold in python
bold = "\033[1m"
regular = "\033[0m"

for i in indices[0]:
  print(f'{bold}Document: {i} {regular}{text[i]}', end='\n\n')
```
