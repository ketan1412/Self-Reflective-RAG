```python
embedding_huggingface = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
vector_store = QdrantVectorStore.from_texts(embedding=embedding_huggingface, collection_name="my_collection_new", texts=docs,
                                            url="URL",
                                            api_key="API_KEY")
```

```python
# selecting top 2 similar documents
response = vector_store.similarity_search_with_score("All about vector database", k=2)

# to make a string bold in python
bold = "\033[1m"
regular = "\033[0m"

# Displaying the documents along with their similarity scores. the higher the score more is the similarity
for doc, score in response:
  print(f"{bold} Document extracted: {regular}{doc.page_content} ,\n {bold} Similarity score: {score}\n\n")
```
